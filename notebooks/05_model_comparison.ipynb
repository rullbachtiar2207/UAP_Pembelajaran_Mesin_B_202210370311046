## 📦 Import Libraries

```python
# ===============================
# Data manipulation & Visualization
# ===============================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
import warnings
warnings.filterwarnings('ignore')

# ===============================
# Visualization Style
# ===============================
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('Set2')

print("✅ Libraries imported successfully!")
```

## 📁 Load Evaluation Results

```python
# Load hasil evaluasi dari 3 model
print("📥 Loading evaluation results...\n")

# LSTM Results
with open('../results/02_lstm_results.json', 'r') as f:
    lstm_results = json.load(f)
print(f"✅ LSTM Results loaded: Accuracy = {lstm_results['accuracy']:.4f}")

# BERT Results
with open('../results/03_bert_results.json', 'r') as f:
    bert_results = json.load(f)
print(f"✅ BERT Results loaded: Accuracy = {bert_results['accuracy']:.4f}")

# DistilBERT Results
with open('../results/04_distilbert_results.json', 'r') as f:
    distilbert_results = json.load(f)
print(f"✅ DistilBERT Results loaded: Accuracy = {distilbert_results['accuracy']:.4f}")

print("\n✅ All results loaded successfully!")
```

## 📊 Comparison Table

```python
# Buat comparison table
comparison_data = [
    {
        'Nama Model': 'LSTM',
        'Akurasi': f"{lstm_results['accuracy']:.4f}",
        'Hasil Analisis': (
            "KELEBIHAN: Model sederhana, cepat training, tidak perlu download model besar.\n"
            "KEKURANGAN: Akurasi lebih rendah, kurang memahami konteks kompleks, perlu data banyak.\n"
            "ALASAN: LSTM adalah model non-pretrained yang harus belajar dari awal, "
            "sehingga butuh data lebih banyak dan waktu lebih lama untuk mencapai performa optimal."
        )
    },
    {
        'Nama Model': 'BERT',
        'Akurasi': f"{bert_results['accuracy']:.4f}",
        'Hasil Analisis': (
            "KELEBIHAN: Akurasi tertinggi, memahami konteks bidirectional, pretrained dengan data besar.\n"
            "KEKURANGAN: Model besar (~110M params), lebih lambat, butuh memori banyak.\n"
            "ALASAN: BERT sudah dilatih dengan 3.3B words, sehingga sudah memiliki pemahaman bahasa yang baik. "
            "Arsitektur bidirectional-nya membuat BERT lebih memahami konteks kalimat."
        )
    },
    {
        'Nama Model': 'DistilBERT',
        'Akurasi': f"{distilbert_results['accuracy']:.4f}",
        'Hasil Analisis': (
            "KELEBIHAN: Balance antara akurasi dan kecepatan, 60% lebih cepat dari BERT, 40% lebih kecil.\n"
            "KEKURANGAN: Akurasi sedikit di bawah BERT, tetap butuh memori cukup besar.\n"
            "ALASAN: DistilBERT adalah versi distilasi dari BERT yang dioptimasi untuk efisiensi. "
            "Menggunakan knowledge distillation, DistilBERT mempertahankan 97% performa BERT dengan resource lebih sedikit."
        )
    }
]

# Convert to DataFrame
comparison_df = pd.DataFrame(comparison_data)

print("="*100)
print("📊 TABEL PERBANDINGAN MODEL")
print("="*100)
print()
print(comparison_df.to_string(index=False))
print()
print("="*100)
```

```python
# Simpan comparison table
comparison_df.to_csv('../results/05_model_comparison.csv', index=False)
print("💾 Comparison table saved: results/05_model_comparison.csv")
```

## 📈 Visualisasi Perbandingan Akurasi

```python
# Bar chart perbandingan akurasi
models = ['LSTM', 'BERT', 'DistilBERT']
accuracies = [
    lstm_results['accuracy'],
    bert_results['accuracy'],
    distilbert_results['accuracy']
]

colors = ['#FF6B6B', '#4ECDC4', '#95E1D3']

plt.figure(figsize=(10, 6))
bars = plt.bar(models, accuracies, color=colors, alpha=0.8, edgecolor='black', linewidth=1.5)

# Add value labels on bars
for i, (bar, acc) in enumerate(zip(bars, accuracies)):
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2., 
        height,
        f'{acc:.4f}\n({acc*100:.2f}%)',
        ha='center', 
        va='bottom',
        fontweight='bold',
        fontsize=11
    )

plt.title('Perbandingan Akurasi Model', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Model', fontsize=12, fontweight='bold')
plt.ylabel('Accuracy', fontsize=12, fontweight='bold')
plt.ylim(0, max(accuracies) * 1.15)
plt.grid(axis='y', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.savefig('../results/05_accuracy_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Accuracy comparison plot saved: results/05_accuracy_comparison.png")
```

## 📊 Training History Comparison

```python
# Load training histories
with open('../results/02_lstm_history.json', 'r') as f:
    lstm_history = json.load(f)

with open('../results/03_bert_history.json', 'r') as f:
    bert_history = json.load(f)

with open('../results/04_distilbert_history.json', 'r') as f:
    distilbert_history = json.load(f)

print("✅ Training histories loaded")
```

```python
# Plot validation accuracy comparison
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Validation Accuracy
axes[0].plot(lstm_history['val_accuracy'], label='LSTM', linewidth=2.5, marker='o', markersize=8)
axes[0].plot(bert_history['val_accuracy'], label='BERT', linewidth=2.5, marker='s', markersize=8)
axes[0].plot(distilbert_history['val_accuracy'], label='DistilBERT', linewidth=2.5, marker='^', markersize=8)
axes[0].set_title('Validation Accuracy Comparison', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Epoch', fontsize=12)
axes[0].set_ylabel('Validation Accuracy', fontsize=12)
axes[0].legend(loc='lower right', fontsize=11)
axes[0].grid(alpha=0.3)

# Validation Loss
axes[1].plot(lstm_history['val_loss'], label='LSTM', linewidth=2.5, marker='o', markersize=8)
axes[1].plot(bert_history['val_loss'], label='BERT', linewidth=2.5, marker='s', markersize=8)
axes[1].plot(distilbert_history['val_loss'], label='DistilBERT', linewidth=2.5, marker='^', markersize=8)
axes[1].set_title('Validation Loss Comparison', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Epoch', fontsize=12)
axes[1].set_ylabel('Validation Loss', fontsize=12)
axes[1].legend(loc='upper right', fontsize=11)
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('../results/05_training_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Training comparison plot saved: results/05_training_comparison.png")
```

## 📊 Detailed Metrics Comparison

```python
# Create detailed comparison table
detailed_comparison = {
    'Model': ['LSTM', 'BERT', 'DistilBERT'],
    'Test Accuracy': [
        f"{lstm_results['accuracy']:.4f}",
        f"{bert_results['accuracy']:.4f}",
        f"{distilbert_results['accuracy']:.4f}"
    ],
    'Best Val Accuracy': [
        f"{lstm_results['training_history']['best_val_accuracy']:.4f}",
        f"{bert_results['training_history']['best_val_accuracy']:.4f}",
        f"{distilbert_results['training_history']['best_val_accuracy']:.4f}"
    ],
    'Best Val Loss': [
        f"{lstm_results['training_history']['best_val_loss']:.4f}",
        f"{bert_results['training_history']['best_val_loss']:.4f}",
        f"{distilbert_results['training_history']['best_val_loss']:.4f}"
    ],
    'Epochs Trained': [
        lstm_results['training_history']['epochs_trained'],
        bert_results['training_history']['epochs_trained'],
        distilbert_results['training_history']['epochs_trained']
    ],
    'Model Type': ['Non-Pretrained', 'Pretrained', 'Pretrained (Distilled)']
}

detailed_df = pd.DataFrame(detailed_comparison)

print("="*100)
print("📊 DETAILED METRICS COMPARISON")
print("="*100)
print()
print(detailed_df.to_string(index=False))
print()
print("="*100)

# Save detailed comparison
detailed_df.to_csv('../results/05_detailed_comparison.csv', index=False)
print("\n💾 Detailed comparison saved: results/05_detailed_comparison.csv")
```

## 🏆 Model Ranking & Recommendation

```python
# Ranking berdasarkan akurasi
rankings = [
    ('LSTM', lstm_results['accuracy']),
    ('BERT', bert_results['accuracy']),
    ('DistilBERT', distilbert_results['accuracy'])
]

rankings.sort(key=lambda x: x[1], reverse=True)

print("="*80)
print("🏆 MODEL RANKING (by Test Accuracy)")
print("="*80)
print()
for i, (model, acc) in enumerate(rankings, 1):
    medal = '🥇' if i == 1 else '🥈' if i == 2 else '🥉'
    print(f"{medal} Rank {i}: {model:12} - Accuracy: {acc:.4f} ({acc*100:.2f}%)")
print()
print("="*80)
```

```python
# Rekomendasi
best_model = rankings[0][0]
best_accuracy = rankings[0][1]

print("\n" + "="*80)
print("💡 REKOMENDASI MODEL")
print("="*80)
print()
print(f"🏆 MODEL TERBAIK: {best_model}")
print(f"📊 Test Accuracy: {best_accuracy:.4f} ({best_accuracy*100:.2f}%)")
print()

if best_model == 'BERT':
    print("✅ BERT memberikan akurasi tertinggi karena:")
    print("   - Pretrained dengan dataset besar (3.3B words)")
    print("   - Arsitektur bidirectional yang powerful")
    print("   - Mampu memahami konteks kompleks dalam teks")
    print()
    print("⚠️ Pertimbangan:")
    print("   - Membutuhkan resource komputasi lebih besar")
    print("   - Waktu inferensi lebih lambat")
    print("   - Jika prioritas adalah KECEPATAN, gunakan DistilBERT")
    print("   - Jika resource terbatas, gunakan LSTM")
    
elif best_model == 'DistilBERT':
    print("✅ DistilBERT memberikan balance terbaik karena:")
    print("   - Akurasi mendekati BERT")
    print("   - 60% lebih cepat dari BERT")
    print("   - Resource requirement lebih rendah")
    print()
    print("💡 Cocok untuk:")
    print("   - Deployment di production")
    print("   - Aplikasi real-time")
    print("   - Resource terbatas")
    
else:  # LSTM
    print("✅ LSTM memberikan performa terbaik karena:")
    print("   - Model sederhana dan cepat")
    print("   - Training time minimal")
    print("   - Resource requirement sangat rendah")
    print()
    print("💡 Cocok untuk:")
    print("   - Baseline model")
    print("   - Resource sangat terbatas")
    print("   - Development cepat")

print("="*80)
```

## 📊 Final Summary

```python
# Buat final summary
final_summary = {
    'comparison_table': comparison_data,
    'rankings': [
        {'rank': i+1, 'model': model, 'accuracy': float(acc)}
        for i, (model, acc) in enumerate(rankings)
    ],
    'best_model': {
        'name': best_model,
        'accuracy': float(best_accuracy),
        'accuracy_percentage': float(best_accuracy * 100)
    },
    'all_models': {
        'lstm': lstm_results,
        'bert': bert_results,
        'distilbert': distilbert_results
    }
}

# Save final summary
with open('../results/05_final_summary.json', 'w') as f:
    json.dump(final_summary, f, indent=4)

print("💾 Final summary saved: results/05_final_summary.json")
```
